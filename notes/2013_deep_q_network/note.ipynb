### Play

```python
if PRAY:
   model_path = "models/dqn_agent_best.pth"
   env = gym.make(game)
   obs = CatcherObserver(env, 84, 84, 4) # width, height, n_frame
   agent = DeepQNetworkAgent.load(actions=list(range(env.action_space.n)), model_path=model_path, device=device)
   agent.play(obs, render=True)
```
