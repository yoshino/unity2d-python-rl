```python
import torch

from code.pendulum_observer import PendulumObserver
from code.ddpg_trainer import DDPGTrainer
from code.ddpg_agent import DDPGAgent
```

```python
device = torch.device("mps" if torch.backends.mps.is_available() else "cpu")
print('device: ', device)
Training = False
```

### Training

```python
if Training:
    observer = PendulumObserver()
    model_path = "models/ddpg_agent.pth"
    trainer = DDPGTrainer(model_path=model_path, device=device)
    agent = trainer.train(env=observer)
```

### Play

```python
observer = PendulumObserver(play=True)
actor_model_path = "models/ddpg_agent_final_actor.pth"
critic_model_path = "models/ddpg_agent_final_critic.pth"
agent = DDPGAgent.load(device=device, state_dim=3, action_dim=observer.action_space, actor_model_path=actor_model_path, critic_model_path=critic_model_path)
agent.play(observer, render=True)
```
