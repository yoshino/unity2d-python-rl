# Double DQN
### リファレンス
[DQNの進化史 ②Double-DQN, Dueling-network, Noisy-network](https://horomary.hatenablog.com/entry/2021/02/06/013412)
> 過大評価は行動選択を、平均値が同じだが別のノイズが乗ったネットワーク(target-Q-networkに対するQ-network)で行うことによって低減することができる、というのがDouble DQNの直感的理解です。
２つのQ-networkでノイズを低減するというDouble DQNの考え方は連続値制御のためのオフポリシー手法であるTD3（DDPGの後継手法）でもclipped-double-Q-learningという名称で採用されています。
